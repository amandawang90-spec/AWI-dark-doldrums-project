```python
import xarray as xr

f = "./solar/data/reconstructed_ssrd/ssrd_reduced_3h_195001-195001.nc"

ds = xr.open_dataset(f)

print(ds)
print()
print("time length:", len(ds.time))
print("variables:", list(ds.data_vars))

x = ds["ssrd"]

print("shape:", x.shape)
print("min:", x.min().compute().item())
print("max:", x.max().compute().item())
print("mean:", x.mean().compute().item())
print("NaN count:", x.isnull().sum().compute().item())
```
